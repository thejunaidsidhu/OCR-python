**`ENGLISH`** OCR


```python
import pytesseract
from pdf2image import convert_from_path

# Convert PDF to images
pdf_path = "decision-tree-algo-ML.pdf"
pages = convert_from_path(pdf_path, dpi=300)

# OCR with pytesseract (English example)
for i, page in enumerate(pages):
    text = pytesseract.image_to_string(page, lang="eng")  # use 'urd' for Urdu
    print(f"\n--- Page {i+1} ---")
    print(text)
```

**`URDU OCR - Different Method`**

```python
from pdf2image import convert_from_path
import os

PDF_FILE = "Shahadat-e-Haq.pdf"
OUTPUT_FOLDER = "pages"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# Convert PDF pages to images
pages = convert_from_path(PDF_FILE, dpi=300)

for idx, page in enumerate(pages):
    image_path = os.path.join(OUTPUT_FOLDER, f"page_{idx+1:03}.png")
    page.save(image_path, "PNG")
    print(f"Saved {image_path}")
```

```python
from pdf2image import convert_from_path
import os

PDF_FILE = "Shahadat-e-Haq.pdf"
OUTPUT_FOLDER = "pages-urdu-pdf"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

pages = convert_from_path(PDF_FILE, dpi=300)

for idx, page in enumerate(pages):
    image_path = os.path.join(OUTPUT_FOLDER, f"page_{idx+1:03}.png")
    page.save(image_path, "PNG")
    print(f"Saved {image_path}")
```

```python
import pytesseract
from PIL import Image
import glob

OUTPUT_TEXT = "urdu_ocr_output.txt"

# Urdu is 'urd'
pytesseract.pytesseract.tesseract_cmd = "/usr/bin/tesseract"  # Default path in Linux

with open(OUTPUT_TEXT, "w", encoding="utf-8") as f:
    for img_path in sorted(glob.glob("pages/*.png")):
        print(f"OCRing {img_path}...")
        text = pytesseract.image_to_string(Image.open(img_path), lang='urd')
        f.write(f"\n\n--- {os.path.basename(img_path)} ---\n")
        f.write(text)

print(f"\nAll OCR complete! Output saved to {OUTPUT_TEXT}")
```
